# Funciones

```python
def comprobacion_por_puntos(df, equipos, jornadas):
    
    for equipo in equipos:
        for jornada in jornadas:
            condicion = (df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)

            PTS_Jugadores = df[condicion]['PTS'].sum()
            PTS_Oficiales = df.loc[condicion, 'RESULTADO'].iloc[0]

            if PTS_Jugadores > PTS_Oficiales:
                print(f'❌ERROR en {equipo}, jornada {jornada}❌ La suma de PTS de los jugadores es mayor a la puntuación oficial')
            elif PTS_Jugadores < PTS_Oficiales:
                print(f'❌ERROR en {equipo}, jornada {jornada}❌ La puntuación oficial es mayor que suma de PTS de los jugadores')
            elif PTS_Jugadores == PTS_Oficiales:
                print(f'✅ La comprobación por puntos ha tenido éxito. {equipo}, jornada {jornada}')
            else:
                print('⚠️ERROR DESCONOCIDO')
```

```python
def comprobar_tiempo(df, equipo, jornada, variable_segundos):
    for equipo in equipos:
        for jornada in jornadas:
            cond1 = df[(df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)][variable_segundos].sum() == (40*60*5*2)
            cond2 = df[(df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)][variable_segundos].sum() == (45*60*5*2)
            cond3 = df[(df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)][variable_segundos].sum() == (50*60*5*2)
            no_prorroga = df[(df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)]['CUARTO'].max() == 4
            una_prorroga = df[(df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)]['CUARTO'].max() == 5
            dos_prorrogas = df[(df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)]['CUARTO'].max() == 6

            if cond1 & no_prorroga:
                print(f'✅ Tiempo correcto. {equipo} - Jornada {jornada}')
            elif cond2 & una_prorroga:
                print(f'✅ Tiempo correcto. {equipo} - Jornada {jornada}')
            elif cond3 & dos_prorrogas:
                print(f'✅ Tiempo correcto. {equipo} - Jornada {jornada}')
            elif no_prorroga:
                print(f'❌ERROR❌ Tiempo incorrecto. {equipo} - Jornada {jornada}: {df[(df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)][variable_segundos].sum()} / {40*60*5*2} segundos')
            elif una_prorroga:
                print(f'❌ERROR❌ Tiempo incorrecto. {equipo} - Jornada {jornada}: {df[(df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)][variable_segundos].sum()} / {45*60*5*2} segundos')
            elif dos_prorrogas:
                print(f'❌ERROR❌ Tiempo incorrecto. {equipo} - Jornada {jornada}: {df[(df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)][variable_segundos].sum()} / {50*60*5*2} segundos')
            else:
                print(f'❌ERROR❌ Error desconocido (Más de dos prórrogas?)')
```

```python
def extraer_dorsal(jugador):
    if len(jugador) > 1 and jugador[1] == ' ':
        return jugador[0]
    else:
        return jugador[:2]
```

```python
def comprobar_totales(df, part1, part2, total):
    check = (df[total].fillna(0) == df[part1] + df[part2]).all()
    if check:
        print('✅Comprobación con ÉXITO')
    else:
        print('⚠️⚠️ALERTA⚠️⚠️. La comprobación ha detectado un error')
```

```python
def resultado_rival(df, equipos, jornadas):
    df['RESULTADO_RIVAL'] = 0
    for equipo in equipos:
        for jornada in jornadas:
            cond_equipo = (df['EQUIPO'] == equipo) & (df['JORNADA'] == jornada)
            rival = df.loc[cond_equipo, 'RIVAL'].iloc[0]
            cond_rival = (df['EQUIPO'] == rival) & (df['JORNADA'] == jornada)

            resultado_rival = df.loc[cond_rival, 'RESULTADO'].iloc[0]
            df['RESULTADO_RIVAL'] = np.where(cond_equipo, resultado_rival, df['RESULTADO_RIVAL'])
```

## Estadísticas de Equipo

- %TR = Porcentaje real de tiro
- %TE = Porcenaje de tiro efectivo
- RT3 = Ratio de triples
- RTL = Ratio de tiros libres
- %RO = Porcenaje de rebotes ofensivos capturados
- %RD = Porcenaje de rebotes defensivos capturados
- %REB = Porcenaje de rebotes capturados
- %ASI = Porcenaje de asistencias
- %ROB = Porcenaje de robos
- %TAP = Porcenaje de tapones
- %PER = Porcenaje de pérdidas
- %USO = Porcenaje de uso

```python
# Eficiencia ofensiva, defensiva y neta del equipo
df['ORtg'] = df['RESULTADO'] * 100 / df['POS_Eq']
df['DRtg'] = df['RESULTADO_RIVAL'] * 100 / df['POS_Rival']
df['NRtg'] = df['ORtg'] - df['DRtg']

# Effective Field Goal (Ajusta el valor extra del Triple)
df['%TE_Eq'] = (df['TCA_Eq'] + 0.5 * df['T3A_Eq']) / df['TCI_Eq']
df['%TE_Eq'] = np.where(df['TCI_Eq'] == 0, 0, df['%TE_Eq'])

# True Shooting % (Mide eficiencia real de anotación incluyendo TL)
df['%TR_Eq'] = df['RESULTADO'] / (2 * (df['TCI_Eq'] + 0.44 * df['TLI_Eq']))
df['%TR_Eq'] = np.where((2 * (df['TCI_Eq'] + 0.44 * df['TLI_Eq'])) == 0, 0, df['%TR_Eq'])

# Assist % (Estimación del porcentaje de canastas asistidas)
df['%ASI_Eq'] = df['ASI_Eq'] / df['TCA_Eq']
df['%ASI_Eq'] = np.where(df['TCA_Eq'] == 0, 0, df['%ASI_Eq'])

# Turnover % (Qué porcentaje de posesiones termina en pérdida)
df['%PER_Eq'] = df['PER_Eq']/df['POS_Eq']
df['%PER_Eq'] = np.where(df['POS_Eq'] == 0, 0, df['%PER_Eq'])

# Offensive Rebound % (Qué porcentaje de rebotes ofensivos posibles captura el equipo)
df['%RO_Eq'] = df['RO_Eq'] / (df['RO_Eq'] + df['RD_Rival'])
df['%RO_Eq'] = np.where((df['RO_Eq'] + df['RD_Rival']) == 0, 0, df['%RO_Eq'])

# Deffensive Rebound % (Qué porcentaje de rebotes defensivos posibles captura el equipo)
df['%RD_Eq'] = df['RD_Eq'] / (df['RD_Eq'] + df['RO_Rival'])
df['%RD_Eq'] = np.where((df['RD_Eq'] + df['RO_Rival']) == 0, 0, df['%RD_Eq'])

# Total Rebound % 
df['%RT_Eq'] = df['RT_Eq'] / (df['RT_Eq'] + df['RT_Rival'])
df['%RT_Eq'] = np.where((df['RT_Eq'] + df['RT_Rival']) == 0, 0, df['%RT_Eq'])

# Ratio de Tiros Libres
df['RTL_Eq'] = df['TLI_Eq'] / df['TCI_Eq']
df['RTL_Eq'] = np.where(df['TCI_Eq'] == 0, 0, df['RTL_Eq'])

# Ratio de Triples Intentados
df['RT3I_Eq'] = df['T3I_Eq'] / df['TCI_Eq']
df['RT3I_Eq'] = np.where(df['TCI_Eq'] == 0, 0, df['RT3I_Eq'])
```

```python
# Porcentaje de tiros de 2, 3 y tiros libres anotados
df['%T2_Eq'] = df['T2A_Eq'] / df['T2I_Eq']
df['%T2_Eq'] = np.where(df['T2I_Eq'] == 0, 0, df['%T2_Eq'])
df['%T3_Eq'] = df['T3A_Eq'] / df['T3I_Eq']
df['%T3_Eq'] = np.where(df['T3I_Eq'] == 0, 0, df['%T3_Eq'])
df['%TL_Eq'] = df['TLA_Eq'] / df['TLI_Eq']
df['%TL_Eq'] = np.where(df['TLI_Eq'] == 0, 0, df['%TL_Eq'])
```

## Estadísticas de Jugador

```python
# % de Uso (Estima qué porcentaje de posesiones del equipo “consume” el jugador mientras está en pista)
    # Interpretación: 30% = jugador ultra dominante        ~20% = rol medio       <15% = jugador de bajo volumen
df['%USO'] = (df['TCI'] + 0.44 * df['TLI'] + df['PER']) * (df['MIN_Eq']) / (df['MIN'] * df['POS_Eq'])
df['%USO'] = np.where((df['MIN'] * df['POS_Eq']) == 0, 0, df['%USO'])

# True Shooting % (Mide eficiencia real de anotación incluyendo TL)
df['%TR'] = df['PTS'] / (2 * (df['TCI'] + 0.44 * df['TLI']))
df['%TR'] = np.where((2 * (df['TCI'] + 0.44 * df['TLI'])) == 0, 0, df['%TR'])

# Effective Field Goal (Ajusta el valor extra del Triple)
df['%TE'] = (df['TCA'] + 0.5 * df['T3A']) / df['TCI']
df['%TE'] = np.where(df['TCI'] == 0, 0, df['%TE'])

# Assist % (Estimación del porcentaje de canastas de compañeros asistidas por el jugador mientras está en pista)
df['%ASI'] = 100 * df['ASI'] / ((df['MIN']/(df['MIN_Eq']) * df['TCA_Eq']) - df['TCA'])
df['%ASI'] = np.where(((df['MIN']/(df['MIN_Eq']) * df['TCA_Eq']) - df['TCA']) == 0, 0, df['%ASI'])

# Turnover % (Qué porcentaje de posesiones usadas termina en pérdida)
df['%PER'] = 100 * df['PER'] / (df['TCI'] + 0.44 * df['TLI'] + df['PER'])
df['%PER'] = np.where((df['TCI'] + 0.44 * df['TLI'] + df['PER']) == 0, 0, df['%PER'])

# Offensive Rebound % (Porcentaje de rebotes ofensivos disponibles capturados)
df['%RO'] = 100 * df['RO'] * (df['MIN_Eq']) / (df['MIN'] * (df['RO_Eq'] + df['RD_Rival']))
df['%RO'] = np.where((df['MIN'] * (df['RO_Eq'] + df['RD_Rival'])) == 0, 0, df['%RO'])

# Deffensive Rebound % (Porcentaje de rebotes defensivos disponibles capturados)
df['%RD'] = 100 * df['RD'] * (df['MIN_Eq']) / (df['MIN'] * (df['RD_Eq'] + df['RO_Rival']))
df['%RD'] = np.where((df['MIN'] * (df['RD_Eq'] + df['RO_Rival'])) == 0, 0, df['%RD'])

# Total Rebound % 
df['%RT'] = 100 * df['RT'] * (df['MIN_Eq']) / (df['MIN'] * (df['RT_Eq'] + df['RT_Rival']))
df['%RT'] = np.where((df['MIN'] * (df['RT_Eq'] + df['RT_Rival'])) == 0, 0, df['%RT'])

# Steal % (Estimación del porcentaje de posesiones rivales robadas)
df['%REC'] = 100 * df['REC'] * (df['MIN_Eq']) / (df['MIN'] * df['POS_Rival'])
df['%REC'] = np.where((df['MIN'] * df['POS_Rival']) == 0, 0, df['%REC'])

# Tapones % (Porcentaje de tiros de 2 rivales taponados)
df['%TAP'] = 100 * df['TAP'] * (df['MIN_Eq']) / (df['MIN'] * (df['T2I_Rival'] - df['T3I_Rival']))
df['%TAP'] = np.where((df['MIN'] * (df['T2I_Rival'] - df['T3I_Rival'])) == 0, 0, df['%TAP'])

# Ratio de Tiros Libres
df['RTL'] = df['TLI'] / df['TCI']
df['RTL'] = np.where(df['TCI'] == 0, 0, df['RTL'])

# Ratio de Triples Intentados
df['RT3I'] = df['T3I'] / df['TCI']
df['RT3I'] = np.where(df['TCI'] == 0, 0, df['RT3I'])

# Puntos por tiro
df['PPT'] = df['PTS'] / df['TCI']
df['PPT'] = np.where(df['TCI'] == 0, 0, df['PPT'])



# # Eficiencia ofensiva individual
# df['TmScPoss'] = df['TCA_Eq'] + (1 - ( 1 - df['TCA_Eq'] / df['TLI_Eq'])**2) * 0.44 * df['TLI_Eq']
# df['TmPoss'] = df['TCI_Eq'] + 0.44 * df['TLI_Eq'] + df['PER_Eq']


# df['qAST'] = df['MIN'] / (df['MIN_Eq']) * (1.14 * (df['ASI_Eq'] - df['ASI'])/(df['TCA_Eq']-df['TCA'])) + ((df['ASI_Eq']/df['MIN_Eq'] * df['MIN'] * 5 - df['ASI']) / ((df['TCA_Eq'] * df['MIN'] * 5) / df['MIN_Eq'] - df['TCA'])) * (1 - df['MIN']/df['MIN_Eq'])
# df['fact_off_Eq'] = df['TmScPoss'] / (df['TCI_Eq'] + 0.44 * df['TLI_Eq'] + df['PER_Eq'])
# df['efic_anot_Eq'] = df['TmScPoss'] / df['TmPoss']
# df['TmROWeight'] = ((1 - df['efic_anot_Eq']) * df['RO_Eq']) / ((1 - df['efic_anot_Eq']) * df['RO_Eq'] + df['RD_Rival'])

# df['pprod_TC'] = 2*(df['T2A'] + 1.5 * df['T3A']) * df['qAST']
# df['pprod_ASI'] = 2*((df['TCA_Eq'] - df['TCA'] + 0.5 * (df['T3A_Eq'] - df['T3A'])) / (df['TCA_Eq'] - df['TCA'])) * df['ASI']
# df['pprod_RO'] = df['RO'] * df['efic_anot_Eq'] * df['TmROWeight']

# df['posesiones_usadas'] = df['TCI'] + 0.44 * df['TLI'] + df['PER']
# df['puntos_producidos'] = df['pprod_TC'] + df['pprod_ASI'] + df['pprod_RO']

# variables_para_ORtg = ['TmScPoss', 'TmPoss', 'qAST', 'fact_off_Eq', 'efic_anot_Eq', 'TmROWeight', 'pprod_TC', 'pprod_ASI', 'pprod_RO', 'posesiones_usadas', 'puntos_producidos']

# df['ORtg'] = 100 * df['puntos_producidos'] / df['posesiones_usadas']


# # Plan B ORtg
# df['pprod_TC'] = 2*(df['T2A'] + 1.5 * df['T3A'])
# df['pprod_ASI'] = 2*((df['TCA_Eq'] - df['TCA'] + 0.5 * (df['T3A_Eq'] - df['T3A'])) / (df['TCA_Eq'] - df['TCA'])) * df['ASI']
# df['pprod_RO'] = df['RO'] * df['efic_anot_Eq'] * df['TmROWeight']

# df['posesiones_usadas'] = df['TCI'] + 0.44 * df['TLI'] + df['PER']
# df['puntos_producidos'] = df['pprod_TC'] + df['pprod_ASI'] + df['pprod_RO']


# df['ORtg2'] = 100 * df['puntos_producidos'] / df['posesiones_usadas']
# df = df.drop(columns=variables_para_ORtg)
```
