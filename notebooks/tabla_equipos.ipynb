```python
import pandas as pd
import numpy as np

import warnings
from pathlib import Path
import unicodedata

pd.set_option("display.max_colwidth", None)



ROOT_PATH = Path(Path.cwd()).resolve().parent
DATA_PATH = ROOT_PATH/'data'/'processed'
PROCESSED_PATH = ROOT_PATH/'data'/'processed'
```

```python
df = pd.read_csv(
    DATA_PATH/'Equipos.csv',
)
```

```python
equipos = []


while True:

    iniciar = input('Añadir un equipo nuevo? (S/N): ').upper()
    if iniciar != 'S':
        break

    nombre = input('Nombre del equipo: ').title()
    codigo = input('Código de tres letras: ').upper()
    ciudad = input('Ciudad: ').title()
    pabellon = input('Nombre del pabellón: ').title()
    aforo = input('Aforo del pabellón: ')
    año_pabellon = input('Año de constucción o remodelación del pabellón: ')
    fundacion = input('Año de fundación del club: ')

    
    nueva_fila = {
        'Codigo': codigo,
        'Nombre': nombre,
        'Ciudad': ciudad,
        'Pabellon': pabellon,
        'Aforo': aforo,
        'Año_Pabellon': año_pabellon,
        'Fundacion': fundacion
    }

    equipos.append(nueva_fila)

df = pd.concat(
    [df, pd.DataFrame(equipos)],
    ignore_index=True
)

```

```python
# ==========================================
# ACTUALIZACIÓN MANUAL DE DATOS DE EQUIPOS
# ==========================================

def preguntar_si_no(pregunta):
    """Pregunta hasta recibir una respuesta válida."""
    while True:
        respuesta = input(f"{pregunta} (s/n): ").strip().lower()

        if respuesta in ["s", "si", "sí"]:
            return True
        elif respuesta in ["n", "no"]:
            return False
        else:
            print("Respuesta no válida. Introduce 's' o 'n'.")


# ==========================================
# 1. CAMBIO DE NOMBRE
# ==========================================

print("\n" + "=" * 50)
print("CAMBIOS DE NOMBRE DE EQUIPO")
print("=" * 50)

while preguntar_si_no("¿Hay algún cambio de nombre?"):

    nombre_antiguo = input("Nombre antiguo: ").strip()
    nombre_actual = input("Nombre actual: ").strip()

    # Comprobar que existe
    if (df["Nombre"] == nombre_antiguo).any():

        df.loc[df["Nombre"] == nombre_antiguo, "Nombre"] = nombre_actual

        print(f"✓ Nombre actualizado: {nombre_antiguo} → {nombre_actual}")

    else:
        print(f"⚠ No se ha encontrado ningún equipo con el nombre '{nombre_antiguo}'.")


# ==========================================
# 2. CAMBIO DE CÓDIGO
# ==========================================

print("\n" + "=" * 50)
print("CAMBIOS DE CÓDIGO")
print("=" * 50)

while preguntar_si_no("¿Hay algún cambio de código?"):

    codigo_antiguo = input("Código antiguo: ").strip().upper()
    codigo_actual = input("Código actual: ").strip().upper()

    # Comprobar que existe
    if (df["Codigo"] == codigo_antiguo).any():

        df.loc[df["Codigo"] == codigo_antiguo, "Codigo"] = codigo_actual

        print(f"✓ Código actualizado: {codigo_antiguo} → {codigo_actual}")

    else:
        print(f"⚠ No se ha encontrado ningún equipo con el código '{codigo_antiguo}'.")


# ==========================================
# 3. NUEVO ESTADIO
# ==========================================

print("\n" + "=" * 50)
print("CAMBIOS DE ESTADIO")
print("=" * 50)

while preguntar_si_no("¿Hay algún cambio de estadio?"):

    codigo = input("Código del equipo: ").strip().upper()

    # Comprobar que existe
    if (df["Codigo"] == codigo).any():

        print("\nIntroduce los nuevos datos del pabellón:")

        pabellon = input("Pabellón: ").strip()

        while True:
            try:
                aforo = int(input("Aforo: "))
                break
            except ValueError:
                print("⚠ El aforo debe ser un número entero.")

        while True:
            try:
                anio = int(input("Año de construcción/inauguración: "))
                break
            except ValueError:
                print("⚠ El año debe ser un número entero.")

        # Actualizar datos
        df.loc[df["Codigo"] == codigo, "Pabellon"] = pabellon
        df.loc[df["Codigo"] == codigo, "Aforo"] = aforo
        df.loc[df["Codigo"] == codigo, "Año_Pabellon"] = anio

        print(f"✓ Estadio actualizado para {codigo}")

    else:
        print(f"⚠ No se ha encontrado ningún equipo con el código '{codigo}'.")

# ==========================================
# 4. EQUIPOS DESCENDIDOS
# ==========================================

print("\n" + "=" * 50)
print("EQUIPOS DESCENDIDOS")
print("=" * 50)

while preguntar_si_no("¿Ha descendido algún equipo?"):

    codigo = input("Código del equipo descendido: ").strip().upper()

    # Comprobar que existe
    if (df["Codigo"] == codigo).any():

        # Guardar nombre para mostrarlo en el mensaje
        nombre = df.loc[df["Codigo"] == codigo, "Nombre"].iloc[0]

        # Eliminar equipo
        df = df[df["Codigo"] != codigo].reset_index(drop=True)

        print(f"✓ Equipo eliminado: {codigo} - {nombre}")

    else:
        print(f"⚠ No se ha encontrado ningún equipo con el código '{codigo}'.")


# ==========================================
# FIN
# ==========================================

print("\n" + "=" * 50)
print("ACTUALIZACIÓN FINALIZADA")
print("=" * 50)
```
