```python
df = pd.read_csv(
    DATA_PATH/'Jugadores.csv',
    dtype={
        "Equipo": "object",
        "Dorsal": "object",
        "Nombre": "object",
        "Apellido": "object",
        "Posición": "object",
        "Altura": "object",
        "Fecha_Nacim": "object",
        "Pais_Nacim": "object",
        "Comunidad": "object",
        "Nacionalidad": "object",
        "Licencia": "object"
    })
```

```python
posiciones = {
    '1': 'Base',
    '2': 'Escolta',
    '3': 'Alero',
    '4': 'Ala-Pívot',
    '5': 'Pívot'
}
licencias = {
    '1': 'JFL',
    '2': 'EUR',
    '3': 'COT',
    '4': 'EXT'
}
```

```python
equipo = input('Equipo: ').upper()
```

```python
jugadores = []

while True:

    continuar = input('Añadir otro jugador? (S/N): ').upper()
    if continuar != 'S':
        break

    # equipo = input('Equipo: ').upper()
    dorsal = input('Dorsal: ')
    nombre = input('Nombre (sin apellido): ').title()
    apellido = input('Apellido: ').title()

    posicion = input('Posición (1=Base, 2=Escolta, 3=Alero, 4=Ala-Pívot, 5=Pívot): ')
    posicion = posiciones.get(posicion, 'Desconocida')

    altura = input('Altura en cm: ')
    dia_nacim = input('Día de nacimiento (DD): ')
    mes_nacim = input('Mes de nacimiento (MM): ')
    año_nacim = input('Año de nacimiento (AAAA): ')

    fecha_nacim = dia_nacim + '/' + mes_nacim + '/' + año_nacim
    
    pais_nacim = input('País de Nacimiento: ').upper()
    if pais_nacim in ['ESP', 'ESPAÑA']:
        comunidad = input('Comunidad Autónoma: ').upper()
    else:
        comunidad = 'NO'

    nacionalidad = input('Nacionalidad = País de Nacimiento? (s/n): ').upper()
    if nacionalidad != 'S':
        nacionalidad = input('Nacionalidad: ').upper()
    else:
        nacionalidad = pais_nacim

    licencia = input('Licencia (1=JFL, 2=EUR, 3=COT, 4=EXT): ').upper()
    if licencia in licencias:
        licencia = licencias.get(licencia, 'Desconocida')
    else:
        licencia = input('Licencia: ').upper()

    nueva_fila = {
        'Equipo': equipo,
        'Dorsal': dorsal,
        'Nombre': nombre,
        'Apellido': apellido,
        'Nombre_Completo': nombre + ' ' + apellido,
        'Posición': posicion,
        'Altura': altura,
        'Fecha_Nacim': fecha_nacim,
        'Pais_Nacim': pais_nacim,
        'Comunidad': comunidad,
        'Nacionalidad': nacionalidad,
        'Licencia': licencia
    }

    jugadores.append(nueva_fila)


df = pd.concat(
    [df, pd.DataFrame(jugadores)],
    ignore_index=True
)
```

```python
iso_paises = {
    'ESPAÑA': 'ESP',
    'ARGENTINA': 'ARG',
    'ITALIA': 'ITA',
    'CHEQUIA': 'CZE',
    'SENEGAL': 'SEN',
    'FRANCIA': 'FRA',
    'GEORGIA': 'GEO',
    'MALI': 'MLI',
    'ESTONIA': 'EST',
    'PAISES BAJOS': 'NLD',
    'SERBIA': 'SRB',
    'CROACIA': 'HRV',
    'HUNGRIA': 'HUN',
    'MONTENEGRO': 'MNE',
    'REINO UNIDO': 'GBR',
    'LITUANIA': 'LTU',
    'SUECIA': 'SWE',
    'CANADA': 'CAN',
    'JAMAICA': 'JAM',
    'URUGUAY': 'URY',
    'RUSIA': 'RUS',
    'CHAD': 'TCD',
    'TURQUIA': 'TUR',
    'REPUBLICA CENTROAFRICANA': 'CAF',
    'REPUBLICA DOMINICANA': 'DOM',
    'CUBA': 'CUB',
    'RUMANIA': 'ROU',
    'ESLOVAQUIA': 'SVK',
    'GUINEA': 'GIN',
    'LETONIA': 'LVA'
}
df.replace(iso_paises, inplace=True)
```

```python
# Comprobación de que todos los países aparecen con código ISO 3166-1 alfa-3

columnas = ["Pais_Nacim", "Comunidad", "Nacionalidad"]

for col in columnas:
    mask = df[col].notna() & (df[col].str.len() > 3)

    if mask.any():
        warnings.warn(
            f"La columna '{col}' contiene valores con más de 3 caracteres: "
            f"{df.loc[mask, col].unique().tolist()}",
            UserWarning
        )
```

```python
df['Apellido_modif'] = (
    df["Apellido"]
    .str.lower()
    .apply(lambda x: unicodedata.normalize("NFD", x).encode("ascii", "ignore").decode("utf-8"))
)

df['Foto'] = 'https://github.com/DJordanV/ACB_26-27/blob/develop/img/clubes/' + df['Equipo'] + '/cuerpo/' + df['Dorsal'] + '-' + df['Apellido_modif'] + '.webp?raw=true'
df['Bandera_Pais'] = 'https://github.com/DJordanV/ACB_26-27/blob/develop/img/banderas/' + df['Pais_Nacim'] + '.png?raw=true'
df['Bandera_Comunidad'] = 'https://github.com/DJordanV/ACB_26-27/blob/develop/img/banderas/ESP/' + df['Comunidad'] + '.png?raw=true'


df = df.drop(columns='Apellido_modif')
```
